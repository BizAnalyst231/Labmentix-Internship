## 3. ***Data Wrangling***

### Data Wrangling Code

````python

# STEP 1: Drop unnecessary column (if exists):
df.drop(columns=['Unnamed: 0'], inplace=True, errors='ignore')
print("\n📌 Dropped 'Unnamed: 0' column!")

# STEP 2: Extract Min and Max Salary from 'Salary Estimate':
# Step 2.1: Remove 'Glassdoor est.' from the 'Salary Estimate' column values-
df['Salary Estimate'] = df['Salary Estimate'].str.replace(r' \(Glassdoor est.\)', '', regex=True)

# Step 2.2: Extract Min and Max Salary from the cleaned 'Salary Estimate' column-

# Extract Min Salary-
df['Min Salary'] = df['Salary Estimate'].str.extract(r'\$(\d+)', expand=False).astype(float) * 1000

# Extract Max Salary-
df['Max Salary'] = df['Salary Estimate'].str.extract(r'-\$(\d+)', expand=False).astype(float) * 1000

# 🔹 STEP 3: Compute Average Salary:
df['Avg Salary'] = (df['Min Salary'] + df['Max Salary']) / 2
print("\n📌 Computed Average Salary:")
print(df[['Min Salary', 'Max Salary', 'Avg Salary']].head())

# 🔹 STEP 4: Clean 'Company Name' by removing ratings:
df['Company Name'] = df['Company Name'].str.replace(r'\n\d+\.\d+', '', regex=True)
print("\n📌 Cleaned 'Company Name' (Removed ratings):")
print(df[['Company Name']].head())

# 🔹 STEP 5: Split 'Location' into 'City' and 'State':
df[['City', 'State']] = df['Location'].str.rsplit(', ', n=1, expand=True)
print("\n📌 Split 'Location' into 'City' and 'State':")
print(df[['Location', 'City', 'State']].head())

# 🔹 STEP 6: Replace '-1' in 'Competitors' with NaN & 'Unknown' in 'Type of ownership':
df['Competitors'] = df['Competitors'].replace('-1', pd.NA)
print("\n📌 Replaced '-1' in 'Competitors' with NaN :")
print(df[['Competitors']].head())
print("\n 'Unknown' in 'Type of ownership:")
df['Type of ownership'] = df['Type of ownership'].replace('-1', 'Unknown')
print(df[['Type of ownership']].head())

# 🔹 STEP 7: Standardize 'Size' and 'Revenue' columns:
df['Size'] = df['Size'].replace('-1', 'Unknown')
df['Revenue'] = df['Revenue'].replace('-1', 'Unknown')
print("\n📌 Standardized 'Size' and 'Revenue' (Replaced '-1' with 'Unknown'):")
print(df[['Size', 'Revenue']].head())

# 🔹 STEP 8: Handle missing values and other attributes:
df['Sector'] = df['Sector'].replace('-1', 'Unknown')  # Replace with 'Unknown'
mean_rating = df[df['Rating'] != -1]['Rating'].mean()  # Compute mean (excluding -1)
df['Rating'] = df['Rating'].replace(-1, mean_rating)   # Replace -1 with mean
df['Industry'] = df['Industry'].replace('-1', 'Unknown')  # Replace '-1' with 'Unknown'
df['Revenue'] = df['Revenue'].astype(str)  # Convert to string type
df['Revenue'].replace(['Unknown / Non-Applicable', 'Unknown'], 'Unknown', inplace=True)
print(df['Revenue'].unique())

# 🔹 STEP 9: Drop duplicates if any:
df.drop_duplicates(inplace=True)
print("\n📌 Dropped duplicate rows (if any).")

# 🔹 STEP 10: Save cleaned data
import os

# Ensure the directory exists:
directory = "D:/Internships/Labmentix"
os.makedirs(directory, exist_ok=True)

cleaned_file_path = os.path.join(directory, "glassdoor_jobs_cleaned.csv")
df.to_csv(cleaned_file_path, index=False)
print("\n✅ Cleaned dataset saved to:", cleaned_file_path)

# 🔹 STEP 11: Display final dataset info:
print("\n📌 Final Dataset Info:")
print(df.info())

# 🔹 STEP 12: Print first 5 rows of cleaned dataset:
print("\n📌 First 5 Rows of Cleaned Data:")
print(df.head())
````

### What all manipulations have you done and insights you found?

Data Cleaning & Processing:

-Dropped unnecessary columns like "Unnamed: 0".

-Extracted Min, Max, and Avg Salary from "Salary Estimate".

-Removed ratings from "Company Name" for better readability.

-Split "Location" into "City" and "State" for granular analysis.

-Replaced "-1" values with NaN or "Unknown" in relevant columns.

-Handled missing values in "Competitors", "Sector", and "Industry".

-Converted revenue values into numerical categories for better visualization.

-Dropped duplicate rows.

Insights Gained:

#### Chart - 6 - DONUT CHART FOR OWNERSHIP TYPES:-

````python
# Count occurrences-
ownership_counts = df['Type of ownership'].value_counts()

# Define figure size-
plt.figure(figsize=(10, 8))

# Define custom colors-
colors = plt.get_cmap('Set3')(range(len(ownership_counts)))

# Create explode effect (only for largest slice)-
explode = [0.1 if i == 0 else 0 for i in range(len(ownership_counts))]

# Function to format percentage labels: Hide for very small values-
def autopct_format(pct):
    return f'{pct:.1f}%' if pct > 2 else ''  # Hide if less than 2%

# Create pie chart-
wedges, texts, autotexts = plt.pie(
    ownership_counts,
    labels=None,  # Remove labels from slices
    autopct=autopct_format,
    startangle=140,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 1.5},
    pctdistance=0.85,
    explode=explode
)

# Improve text readability-
for text in autotexts:
    text.set_fontsize(12)

# Add a circle to create a "donut chart" effect-
centre_circle = plt.Circle((0, 0), 0.60, fc='white')
plt.gca().add_patch(centre_circle)

# Add title-
plt.title('Type of Ownership Distribution', fontsize=14)

# Add legend instead of labels on the pie-
plt.legend(wedges, ownership_counts.index, title="Ownership Types", loc="best", bbox_to_anchor=(1, 0.5))

# Show plot-
plt.show()
````

 To understand company ownership distribution.
